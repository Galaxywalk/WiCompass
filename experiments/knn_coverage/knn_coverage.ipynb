# mmBody Cross-Dataset Coverage

```python
# -*- coding: utf-8 -*-
# Plot: k vs coverage rate
# Reads data from JSON files in logs/wicompass/knn_coverage/mmbody/
# Legend inside figure, right-middle

import json
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

# ---- Style ----
color_A = '#1e90ff'   # Blue - mmBody -> AMASS
color_B = '#ffbb00'   # Yellow - AMASS -> mmBody
marker_A = 'o'
marker_B = 's'

plt.rcdefaults()
plt.rcParams["figure.figsize"] = [5, 3]
plt.rcParams["figure.dpi"] = 100
plt.rcParams["grid.linestyle"] = "--"
plt.rcParams['lines.linewidth'] = 2.5
plt.rcParams['lines.markersize'] = 8
plt.rcParams['lines.markeredgewidth'] = 2
plt.rcParams["font.size"] = 16
plt.rcParams["font.family"] = "Arial"
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42
plt.rcParams["legend.fontsize"] = 10
plt.rcParams['legend.frameon'] = False

# ---- Load data from JSON files ----
data_dir = Path("../../logs/wicompass/knn_coverage/mmbody")
selected_k = [2, 4, 6, 8, 10, 12]

def load_coverage_from_json(k: int) -> tuple:
    """
    Load coverage rates from JSON file for given k.
    Returns (cov_A_to_B, cov_B_to_A), or (0, 0) if file doesn't exist.
    """
    json_path = data_dir / f"k{k}_dedup.json"
    if not json_path.exists():
        print(f"Warning: {json_path} not found, using 0")
        return 0.0, 0.0
    
    with open(json_path, 'r') as f:
        data = json.load(f)
    
    # Coverage = 1 - hole_rate
    hole_A_to_B = data["A_to_B"]["hole_rate"]
    hole_B_to_A = data["B_to_A"]["hole_rate"]
    
    return 1 - hole_A_to_B, 1 - hole_B_to_A

# Load coverage for all selected k values
cov_A_to_B = []
cov_B_to_A = []

for k in selected_k:
    cov_a, cov_b = load_coverage_from_json(k)
    cov_A_to_B.append(cov_a)
    cov_B_to_A.append(cov_b)

print(f"Selected k values: {selected_k}")
print(f"Coverage A->B (mmBody->AMASS): {[f'{c:.4f}' for c in cov_A_to_B]}")
print(f"Coverage B->A (AMASS->mmBody): {[f'{c:.4f}' for c in cov_B_to_A]}")

# ---- Plot ----
fig, ax = plt.subplots()

line1, = ax.plot(selected_k, cov_A_to_B, color=color_A, marker=marker_A,
                 label=r"$\mathrm{Coverage}(\Phi_{\mathrm{mmBody}} \to \Phi_{\mathrm{AMASS}})$")
line2, = ax.plot(selected_k, cov_B_to_A, color=color_B, marker=marker_B,
                 label=r"$\mathrm{Coverage}(\Phi_{\mathrm{AMASS}} \to \Phi_{\mathrm{mmBody}})$")

# Value annotations
for x, y in zip(selected_k, cov_A_to_B):
    ax.text(x, y + 0.045, f"{y*100:.1f}%", ha="center", va="bottom", fontsize=12, color=color_A)
for x, y in zip(selected_k, cov_B_to_A):
    ax.text(x, y + 0.045, f"{y*100:.1f}%", ha="center", va="bottom", fontsize=12, color=color_B)

# ax.set_xlabel("k")
ax.set_ylabel("Coverage rate")
ax.set_ylim(0, 1.2)           # Slightly above 100% for annotation space
ax.set_xticks(selected_k)     # Only show selected k values on x-axis
ax.yaxis.set_major_formatter(PercentFormatter(1.0))
ax.grid(True, linestyle="--", alpha=0.6)

# Legend inside figure, center-right
ax.legend(loc="center", bbox_to_anchor=(0.6, 0.45))

plt.tight_layout()
plt.savefig("mmbody_coverage_rate_vs_k.pdf", bbox_inches="tight")

plt.show()
```
